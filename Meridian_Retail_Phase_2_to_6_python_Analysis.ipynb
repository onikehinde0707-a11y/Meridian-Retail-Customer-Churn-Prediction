## 2.2 Entity-Relationship Structure

The five datasets are connected through common identifiers that describe how customers, orders, products, and support interactions relate to one another.

- Customers -> Orders: One customer can have many orders, linked through 'customer_id'.
- Orders -> Order Items: One order can contain many order items, linked through 'order_id'.
- Products -> Order Items: One product can aappear in many order items, linked through 'product_id'.
- Customers -> Support Tickets: One customer can have many support tickets, linked through 'customer_id'.

These relationships allow customer information, purchasing behaviour, product information, and customer-service interactions to be combined when investigating customer churn.

```python
from graphviz import Digraph

er = Digraph("Meridian_ER", format="png")

# Tables
er.node("Customers", "CUSTOMERS\ncustomer_id (PK)" )
er.node("Orders", "ORDERS\norder_id (PK)\ncustomer_id (FK)" )
er.node("Order Items", "ORDER ITEMS\norder_id (FK)\nproduct_id (FK)\nquantity" )
er.node("Products", "PRODUCTS\nproduct_id (PK)" )
er.node("Support Tickets", "SUPPORT TICKETS\nticket_id (PK)\ncustomer_id (FK)")

# Relationships
er.edge("Customers", "Orders", label="1 : Many" )
er.edge("Orders", "Order_Items", label="1 : Many" )
er.edge("Products", "Order_Items", label="1 : Many")
er.edge("Customers", "Support Tickets", label="1 : Many")

er
```

What the terms mean
- PK = Primary key - uniquely identifies a record.
- FK = Foreign key - connects one table to another.
- 1 : Many = one record in the first table can be connected to many records in the second table.

```python
# Convert customer signup dates to proper datetime format

customers["signup_date"] = pd.to_datetime(
    customers["signup_date"],
    errors="coerce"
)

print(customers["signup_date"].dtype)
print("Missing signup dates:", customers["signup_date"].isna().sum())
```

```python
# Convert order dates to proper datetime format

orders["order_date"] = pd.to_datetime(
    orders["order_date"],
    errors="coerce"
)

print(orders["order_date"].dtype)
print("Missing order dates:", orders["order_date"].isna().sum())
```

```python
# Convert product unit prices to numeric values

products["unit_price"] = (
    products["unit_price"]
    .astype(str)
    .str.replace("$", "", regex=False)
    .str.replace(",", "", regex=False)
    .str.strip()
)

products["unit_price"] = pd.to_numeric(
    products["unit_price"],
    errors="coerce"
)

print(products["unit_price"].dtype)
print("Missing product prices:", products["unit_price"].isna().sum())
```

Explanation: I cleaned the product unit prices by removing currency symbols, commas and unnecessary spaces, then converted the values to a numeric format using Pandas. I used errors="coerce" to safely handle any invalid price values. The result showed that the prices are now stored as float64, with 0 missing product prices after cleaning.

```python
# Check missing purchase prices and related information

missing_purchase_price = order_items[
    order_items["unit_price_at_purchase"].isna()
]

print("Rows with missing purchase price:", len(missing_purchase_price))
print("\nSample of affected rows:")
print(
    missing_purchase_price[
        ["order_item_id", "order_id", "product_id", "quantity"]
    ].head(10)
)
```

Explanation: I identified the order-item records with missing purchase prices and reviewed their related order IDs, product IDs and quantities. There were 1,469 records with missing purchase prices. The affected records still contained valid order and product identifiers, so I investigated whether their prices could be matched to the Products dataset rather than replacing the missing values with zero.

```python
# Check whether products have prices for the affected order items

price_check = missing_purchase_price.merge(
    products[["product_id", "unit_price"]],
    on="product_id",
    how="left"
)

print("Missing purchase prices:", len(price_check))
print("Missing matching product prices:", price_check["unit_price"].isna().sum())

print("\nSample of matched product prices:")
print(price_check.head(10))
```

Explanation: I matched the order items with missing purchase prices to the Products dataset using product_id. All 1,469 affected records had a matching product price, with 0 missing matching product prices. This confirmed that the Products dataset could be used as a reliable fallback for filling the missing purchase prices.

```python
# Fill missing purchase prices using the matching product price

order_items = order_items.merge(
    products[["product_id", "unit_price"]].rename(
        columns={"unit_price": "product_unit_price"}
    ),
    on="product_id",
    how="left"
)

order_items["unit_price_at_purchase"] = (
    order_items["unit_price_at_purchase"]
    .fillna(order_items["product_unit_price"])
)

order_items.drop(columns=["product_unit_price"], inplace=True)

print(
    "Missing purchase prices after cleaning:",
    order_items["unit_price_at_purchase"].isna().sum()
)
```

Explanation: I filled the missing unit_price_at_purchase values using the corresponding product price from the Products dataset. This was possible because all 1,469 affected records had a matching product price. After the cleaning, the number of missing purchase prices reduced to 0.

```python
# Check products with missing sub-categories

missing_subcategory = products[
    products["sub_category"].isna()
]

print("Missing sub-categories:", len(missing_subcategory))
print("\nAffected products:")
print(
    missing_subcategory[
        ["product_id", "product_name", "category", "brand"]
    ]
)
```

Explanation: I identified the products with missing sub-category information and reviewed their product names, categories and brands. There were 11 products with missing sub-categories. I did not assign values immediately because I wanted to check whether similar products from the same category and brand could provide a reliable basis for completing the missing information.

```python
# Find products with the same category and brand as the affected products

for _, row in missing_subcategory.iterrows():
    matches = products[
        (products["category"] == row["category"]) &
        (products["brand"] == row["brand"]) &
        (products["sub_category"].notna())
    ]

    print(f"\n{row['product_id']} - {row['product_name']}")
    print(matches[["product_name", "sub_category"]].head())
```

Explanation: I checked the Products dataset for missing sub-category information and reviewed the affected products along with their product names, categories and brands. There were 11 products with missing sub-categories. I did not fill the missing values immediately because I wanted to first investigate the affected products and compare them with similar products in the same category and brand. This helped me choose suitable sub-categories based on the available data rather than making unsupported assumptions.

```python
# Fill the 11 missing product sub-categories

subcategory_updates = {
    "PROD1030": "Team Sports",
    "PROD1069": "Small Appliances",
    "PROD1101": "Camping",
    "PROD1116": "Fragrance",
    "PROD1117": "Men's Clothing",
    "PROD1130": "Haircare",
    "PROD1161": "Skincare",
    "PROD1162": "Smart Home",
    "PROD1167": "Tablets",
    "PROD1205": "Cameras",
    "PROD1206": "Laptops"
}

products["sub_category"] = products["product_id"].map(
    subcategory_updates
).fillna(products["sub_category"])

print(
    "Missing sub-categories after cleaning:",
    products["sub_category"].isna().sum()
)
```

Explanation: I filled the 11 missing product sub-categories using the product information and the investigation carried out in the previous step. After cleaning, the number of missing sub-categories reduced from 11 to 0, showing that all missing sub-category values were successfully completed.

```python
# Compare duplicate customer IDs with their signup dates

confirmed_duplicate_pairs = (
    customers.assign(
        email_normalized=customers["email"].str.lower()
    )
    .dropna(subset=["email_normalized"])
    .groupby("email_normalized")
    .filter(lambda x: len(x) > 1)
    .sort_values("email_normalized")
)

print(
    confirmed_duplicate_pairs[
        ["customer_id", "first_name", "last_name",
         "email", "signup_date"]
    ].head(30).to_string(index=False)
)
```

I will retain the original customer record and carefully reassign any related transactions from the duplicate ID before removing the duplicate record.

```python
# Identify duplicate customer pairs and compare their customer IDs

duplicate_pairs_check = (
    customers.assign(
        email_normalized=customers["email"].str.lower()
    )
    .dropna(subset=["email_normalized"])
    .groupby("email_normalized")
    .filter(lambda x: len(x) > 1)
)

duplicate_pairs_check["id_number"] = (
    duplicate_pairs_check["customer_id"]
    .str.extract(r"(\d+)")
    .astype(int)
)

print("Total customer records involved in confirmed duplicate emails:",
      len(duplicate_pairs_check))

print("Unique duplicate email groups:",
      duplicate_pairs_check["email_normalized"].nunique())

print("\nHighest customer IDs involved:")
print(
    duplicate_pairs_check[
        ["customer_id", "email_normalized", "id_number"]
    ]
    .sort_values("id_number", ascending=False)
    .head(20)
    .to_string(index=False)
)
```

Explanation: I confirmed that 176 customer records were involved in repeated email addresses, representing 88 duplicate email groups. The duplicate records are concentrated among the higher customer IDs, with IDs such as CUST006090, CUST006089, and CUST006081. This pattern supports the finding that these customers were re-entered under new customer IDs. I will continue investigating their related orders and support tickets before consolidating the duplicate records, so that no customer activity is lost.

```python
# Check orders and support tickets linked to duplicate customer IDs

duplicate_customer_ids = duplicate_pairs_check["customer_id"].unique()

duplicate_orders_count = orders[
    orders["customer_id"].isin(duplicate_customer_ids)
].shape[0]

duplicate_tickets_count = support_tickets[
    support_tickets["customer_id"].isin(duplicate_customer_ids)
].shape[0]

print("Orders linked to duplicate customer IDs:", duplicate_orders_count)
print("Support tickets linked to duplicate customer IDs:", duplicate_tickets_count)
```

Explanation: I checked whether the confirmed duplicate customer IDs were linked to transaction and support records. The results showed that 791 orders and 163 support tickets were associated with the duplicate customer IDs. This confirms that the duplicate customer records have related business activity. Therefore, the duplicate customer IDs cannot be deleted directly; their orders and support tickets must first be reassigned to the corresponding original customer IDs to preserve the customer’s transaction and service history.

```python
# Create a mapping from duplicate customer IDs to original customer IDs

duplicate_customer_mapping = {}

for email, group in (
    duplicate_pairs_check
    .groupby("email_normalized")
):
    customer_ids = group["customer_id"].tolist()

    # Keep the lower customer ID as the original record
    original_id = min(customer_ids)

    for customer_id in customer_ids:
        if customer_id != original_id:
            duplicate_customer_mapping[customer_id] = original_id

print("Duplicate customer IDs to be reassigned:",
      len(duplicate_customer_mapping))

print("\nSample ID mapping:")
for duplicate_id, original_id in list(
    duplicate_customer_mapping.items()
)[:20]:
    print(duplicate_id, "->", original_id)
```

Explanation: I created a mapping between the duplicate customer IDs and their corresponding original customer IDs. The analysis identified 88 duplicate customer IDs that need to be reassigned. I used the lower customer ID as the original record because the duplicate records were re-entered under newer, higher IDs. This mapping will allow the related orders and support tickets to be transferred to the original customer records before the duplicate customer records are removed, preventing loss of transaction and support history.

```python
# Reassign orders from duplicate customer IDs to original customer IDs

orders["customer_id"] = (
    orders["customer_id"]
    .replace(duplicate_customer_mapping)
)

remaining_duplicate_orders = orders[
    orders["customer_id"].isin(duplicate_customer_mapping.keys())
].shape[0]

print(
    "Orders still linked to duplicate customer IDs:",
    remaining_duplicate_orders
)
```

Explanation: I reassigned the orders linked to the duplicate customer IDs to their corresponding original customer IDs using the duplicate-customer mapping. After the reassignment, I checked for any orders that were still linked to duplicate IDs. The result was 0, confirming that all 791 affected orders were successfully reassigned without losing their transaction records.

```python
# Reassign support tickets from duplicate customer IDs to original customer IDs

support_tickets["customer_id"] = (
    support_tickets["customer_id"]
    .replace(duplicate_customer_mapping)
)

remaining_duplicate_tickets = support_tickets[
    support_tickets["customer_id"].isin(duplicate_customer_mapping.keys())
].shape[0]

print(
    "Support tickets still linked to duplicate customer IDs:",
    remaining_duplicate_tickets
)
```

Explanation: I reassigned the support tickets linked to the duplicate customer IDs to their corresponding original customer IDs using the duplicate-customer mapping. After the reassignment, I checked for any support tickets that were still linked to duplicate IDs. The result was 0, confirming that all 163 affected support tickets were successfully reassigned while preserving the customers’ support history.

```python
# Final check for references to duplicate customer IDs

duplicate_ids = set(duplicate_customer_mapping.keys())

orders_still_linked = orders[
    orders["customer_id"].isin(duplicate_ids)
].shape[0]

tickets_still_linked = support_tickets[
    support_tickets["customer_id"].isin(duplicate_ids)
].shape[0]

customers_still_present = customers[
    customers["customer_id"].isin(duplicate_ids)
].shape[0]

print("Orders still linked to duplicate IDs:", orders_still_linked)
print("Support tickets still linked to duplicate IDs:", tickets_still_linked)
print("Duplicate customer records still present:", customers_still_present)
```

The 88 duplicate customer records were still present in the Customers dataset, as expected, because they had not yet been removed. This confirmed that the related transaction and support records had been successfully reassigned and that the duplicate customer records could now be safely removed.

```python
# Remove confirmed duplicate customer records

customers = customers[
    ~customers["customer_id"].isin(duplicate_customer_mapping.keys())
].reset_index(drop=True)

print("Customers after removing duplicates:", len(customers))

print(
    "Duplicate customer IDs still present:",
    customers["customer_id"].isin(
        duplicate_customer_mapping.keys()
    ).sum()
)
```

```python
# Check for outliers in order total amounts using the IQR method

Q1 = orders["total_amount"].quantile(0.25)
Q3 = orders["total_amount"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers_total_amount = orders[
    (orders["total_amount"] < lower_bound) |
    (orders["total_amount"] > upper_bound)
]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower bound:", lower_bound)
print("Upper bound:", upper_bound)
print("Number of outliers:", len(outliers_total_amount))
```

I used the Interquartile Range (IQR) method to identify potential outliers in the total_amount column of the Orders dataset. The calculated upper bound was 4,306.31, and 862 orders were identified as potential outliers. These records were retained because unusually high order amounts may represent legitimate high-value purchases rather than data errors. Removing them without further evidence could lead to the loss of important sales information.
