Note: Validating the Analysis Reference Date

The project data dictionary specifies 1 July 2026 as the reference date for recency analysis. I found two orders dated after this date: one returned order and one completed order. To maintain a consistent observation period, orders dated after 1 July 2026 will be excluded from the churn calculations. The original dataset remains unchanged.

```python
# PHASE 7: PREPARE VALID COMPLETED ORDERS

reference_date = pd.Timestamp("2026-07-01")

completed_orders_valid = completed_orders[
    (completed_orders["order_date"].notna()) &
    (completed_orders["order_date"] <= reference_date)
].copy()

print("Completed orders before filtering:", len(completed_orders))
print("Valid completed orders for analysis:", len(completed_orders_valid))
print("Orders excluded due to missing dates:",
      completed_orders["order_date"].isna().sum())
print("Orders excluded after reference date:",
      (completed_orders["order_date"] > reference_date).sum())
```

Note: Preparing Completed Orders

I filtered the completed-order dataset to retain records with valid order dates on or before 1 July 2026. Of the original 25,481 completed orders, 25,096 met these criteria. I excluded 384 orders with missing dates and one order dated after the reference date from this working dataset. This ensures consistent recency calculations without deleting records from the original dataset.

```python
# PHASE 7: CALCULATE CUSTOMER RECENCY

# Find each customer's most recent completed order
last_purchase = (
    completed_orders_valid
    .groupby("customer_id")["order_date"]
    .max()
    .reset_index()
)

# Rename the column for clarity
last_purchase.rename(
    columns={"order_date": "last_purchase_date"},
    inplace=True
)

# Calculate days since the last purchase
last_purchase["recency_days"] = (
    reference_date - last_purchase["last_purchase_date"]
).dt.days

print("Customers with at least one valid completed order:",
      len(last_purchase))

print("\nRecency summary:")
print(last_purchase["recency_days"].describe())
```

Note: Calculating Customer Recency

I calculated recency as the number of days between each customer’s most recent valid completed order and the reference date of 1 July 2026. The analysis identified 5,584 customers with at least one valid completed order. Their average recency was approximately 362 days, while the median was 297 days. These results suggest that customer inactivity warrants further investigation using the project’s 180-day churn threshold. Customers without valid completed orders will be accounted for separately in the customer-level analysis.

```python
# PHASE 7: CALCULATE CUSTOMER PURCHASE FREQUENCY

# Count valid completed orders for each customer
purchase_frequency = (
    completed_orders_valid
    .groupby("customer_id")
    .size()
    .reset_index(name="frequency")
)

print("Customers with completed orders:",
      len(purchase_frequency))

print("\nPurchase frequency summary:")
print(purchase_frequency["frequency"].describe())
```

Purchase Frequency Analysis

I calculated purchase frequency by counting each customer’s valid completed orders on or before 1 July 2026. The analysis covered 5,584 customers, with an average of approximately 4.49 completed orders per customer. The median was 4 orders, while the number of completed orders ranged from 1 to 16. These results show differences in customer purchasing behaviour and will help us investigate the relationship between purchase frequency and churn.

```python
# PHASE 7: CALCULATE CUSTOMER MONETARY VALUE

# Calculate total spending for each customer
monetary_value = (
    completed_orders_valid
    .groupby("customer_id")["total_amount"]
    .sum()
    .reset_index(name="monetary_value")
)

print("Customers with completed orders:",
      len(monetary_value))

print("\nMonetary value summary:")
print(monetary_value["monetary_value"].describe().round(2))
```

Monetary Value Analysis

I calculated the total monetary value for each customer by summing the amounts of their valid completed orders on or before 1 July 2026. The analysis covered 5,584 customers. Average total spending was 6,914.99, while median spending was 6,352.73. Total spending ranged from 25.88 to 35,395.37. The average being higher than the median suggests that some customers have relatively high total spending. Further analysis will examine whether spending patterns differ between active and churned customers.

```python
# PHASE 7: COMBINE CUSTOMER RFM FEATURES

# Start with customers who have a valid completed order
rfm = (
    last_purchase[
        ["customer_id", "last_purchase_date", "recency_days"]
    ]
    .merge(purchase_frequency, on="customer_id", how="left")
    .merge(monetary_value, on="customer_id", how="left")
)

# Display the first five rows
print("RFM table preview:")
display(rfm.head())

# Check the number of customers and columns
print("\nRFM table shape:", rfm.shape)

# Check for missing values
print("\nMissing values:")
print(rfm.isnull().sum())
```

RFM Feature Creation

I combined recency, frequency and monetary value into a single customer-level table. The table contains 5,584 customers and five columns: customer ID, last purchase date, recency in days, purchase frequency and total monetary value. No missing values were found in these five columns. The RFM features will help identify differences in customer purchasing behaviour and support the investigation of churn patterns. A further 418 customers were not represented because they had no valid completed order in the current analysis; these customers will be considered separately to avoid excluding them from the overall churn assessment.

```python
# PHASE 7: INVESTIGATE CUSTOMERS WITHOUT VALID COMPLETED ORDERS

# Identify customers missing from the RFM table
customers_without_rfm = customers[
    ~customers["customer_id"].isin(rfm["customer_id"])
].copy()

# Find all orders belonging to these customers
orders_without_rfm = orders[
    orders["customer_id"].isin(
        customers_without_rfm["customer_id"]
    )
].copy()

print("Customers without an RFM record:",
      customers_without_rfm["customer_id"].nunique())

print("\nOrder status for these customers:")
print(orders_without_rfm["order_status"].value_counts(dropna=False))

print("\nTotal orders belonging to these customers:",
      len(orders_without_rfm))

print("\nOrders with missing dates:",
      orders_without_rfm["order_date"].isna().sum())
```

Customer ID Matching Validation

I compared the customer IDs in the RFM table with those in the customer dataset. The check identified 18 RFM records whose customer IDs were not present in the customer table.
```python
# PHASE 7: INVESTIGATE UNMATCHED CUSTOMER IDs

# Identify customer IDs present in RFM but absent from customers
unmatched_ids = rfm_ids_not_in_customers["customer_id"].tolist()

# Check whether these IDs appear in the orders dataset
unmatched_order_records = orders[
    orders["customer_id"].isin(unmatched_ids)
]

print("Unmatched customer IDs:", len(unmatched_ids))

print("\nOrders linked to unmatched IDs:",
      len(unmatched_order_records))

print("\nNumber of distinct unmatched IDs in orders:",
      unmatched_order_records["customer_id"].nunique())

print("\nOrder status for unmatched IDs:")
print(unmatched_order_records["order_status"].value_counts())
```

Investigation of Unmatched Customer IDs

I investigated the 18 customer IDs present in the RFM table but absent from the customer dataset. All 18 IDs were found in the orders dataset, and each was associated with one completed order. This indicates that these records have purchase activity but lack corresponding customer profiles. I will exclude these unmatched IDs from customer-level analyses that require verified customer information and document them as a data quality limitation.

```python
# PHASE 7: CREATE VERIFIED CUSTOMER-LEVEL RFM TABLE

# Keep only RFM records linked to valid customer records
rfm_verified = rfm[
    rfm["customer_id"].isin(customers["customer_id"])
].copy()

# Add customer information to the RFM table
customer_rfm = customers.merge(
    rfm_verified,
    on="customer_id",
    how="left"
)

print("Total verified customers:", len(customer_rfm))

print("Unique customer IDs:",
      customer_rfm["customer_id"].nunique())

print("\nCustomers with RFM records:",
      customer_rfm["recency_days"].notna().sum())

print("Customers without RFM records:",
      customer_rfm["recency_days"].isna().sum())

print("\nCombined table shape:", customer_rfm.shape)
```

Creation of the Verified Customer-Level RFM Table

I combined the cleaned customer dataset with the RFM features using customer ID as the matching key. The resulting table contains 6,002 unique customers and 24 columns. Of these, 5,566 customers have verified RFM records, while 436 have no qualifying RFM record. I excluded 18 unmatched RFM records because their customer IDs were absent from the customer dataset. This verified table provides a consistent customer-level dataset for the subsequent churn analysis.

```python
# PHASE 7: DEFINE CUSTOMER CHURN

# Set the churn inactivity threshold
churn_threshold = 180

# Classify customers with a valid last purchase date
customer_rfm["churn_status"] = pd.NA

customer_rfm.loc[
    customer_rfm["recency_days"] > churn_threshold,
    "churn_status"
] = 1

customer_rfm.loc[
    customer_rfm["recency_days"] <= churn_threshold,
    "churn_status"
] = 0

# Identify customers without a valid RFM record
customer_rfm["no_valid_rfm_record"] = (
    customer_rfm["recency_days"].isna()
)

print("Churn threshold:", churn_threshold, "days")

print("\nChurn classification:")
print(
    customer_rfm["churn_status"]
    .value_counts(dropna=False)
    .rename(index={1: "Churned", 0: "Not churned"})
)

print("\nCustomers without valid RFM records:",
      customer_rfm["no_valid_rfm_record"].sum())
```

Initial Customer Churn Classification

I classified customers using a 180-day inactivity threshold and 1 July 2026 as the reference date. Customers whose last valid completed order was more than 180 days before the reference date were labelled as churned, while customers whose last valid completed order was within the previous 180 days were labelled as not churned. The initial classification identified 3,565 churned customers and 2,001 customers who were not churned. A further 436 customers remained unclassified because they had no valid RFM record. The initial churn rate among the 5,566 classifiable customers was 64.05%. This figure will be treated as provisional until the records of the unclassified customers are investigated.

The initial churn rate among customers who can be classified is:

\text{Churn Rate}=\frac{3,565}{3,565+2,001}\times100

Initial churn rate = 64.05%.

```python
# PHASE 7: INVESTIGATE UNCLASSIFIED CUSTOMERS

# Identify customers without a valid RFM record
unclassified_customers = customer_rfm[
    customer_rfm["no_valid_rfm_record"]
].copy()

# Check whether these customers have any orders
unclassified_orders = orders[
    orders["customer_id"].isin(
        unclassified_customers["customer_id"]
    )
].copy()

print("Unclassified customers:",
      unclassified_customers["customer_id"].nunique())

print("\nCustomers with any recorded orders:",
      unclassified_orders["customer_id"].nunique())

print("\nOrder status:")
print(unclassified_orders["order_status"].value_counts(dropna=False))

print("\nOrders with missing dates:",
      unclassified_orders["order_date"].isna().sum())
```

Investigation of Unclassified Customers

The analysis identified 436 customers without a valid RFM record. Of these, 18 had recorded orders: 11 returned, 7 cancelled, 5 pending, and 2 completed. Three orders had missing dates. The remaining 418 customers had no matching orders.

These customers were left unclassified because there was insufficient reliable information to determine their last valid completed purchase date. Excluding them from the churn-rate denominator avoids assigning an unsupported churn status.

```python
# PHASE 7: CHURN RATE BY MEMBERSHIP TIER

# Keep only customers with a known churn classification
classified_customers = customer_rfm[
    customer_rfm["churn_status"].notna()
].copy()

# Calculate customer counts and churn rates by membership tier
churn_by_tier = (
    classified_customers
    .groupby("membership_tier")
    .agg(
        total_customers=("customer_id", "count"),
        churned_customers=("churn_status", "sum")
    )
    .reset_index()
)

# Calculate churn rate as a percentage
churn_by_tier["churn_rate_pct"] = (
    churn_by_tier["churned_customers"]
    / churn_by_tier["total_customers"]
    * 100
).round(2)

# Display the results
print("Churn rate by membership tier:")
display(churn_by_tier)
```

Churn Rate by Membership Tier

The analysis examined churn rates among customers with valid churn classifications across four membership tiers. Gold members had the highest churn rate at 65.14%, followed by Silver at 63.98%, Bronze at 63.84%, and Platinum at 62.80%.

The difference between the highest and lowest rates was approximately 2.33 percentage points. This indicates that churn was relatively high across all membership tiers. However, membership tier alone may not explain customer churn, and further analysis is required to investigate other contributing factors.

Limitation: The calculation excludes 436 customers without a valid RFM record. Therefore, the rates represent classified customers rather than the entire customer base.

```python
# PHASE 7: CHURN RATE BY PREFERRED CHANNEL

# Keep customers with a known churn classification
classified_customers = customer_rfm[
    customer_rfm["churn_status"].notna()
].copy()

# Group customers by their preferred shopping channel
churn_by_channel = (
    classified_customers
    .groupby("preferred_channel", dropna=False)
    .agg(
        total_customers=("customer_id", "count"),
        churned_customers=("churn_status", "sum")
    )
    .reset_index()
)

# Calculate churn rate as a percentage
churn_by_channel["churn_rate_pct"] = (
    churn_by_channel["churned_customers"]
    / churn_by_channel["total_customers"]
    * 100
).round(2)

# Display the results
print("Churn rate by preferred channel:")
display(churn_by_channel)
```

Churn Rate by Preferred Channel

The analysis compared churn rates among customers with valid churn classifications across three preferred shopping channels. In-Store customers had the highest churn rate at 65.29%, followed by Mobile App customers at 63.78%. Online customers had the lowest churn rate at 63.07%.

```python
# PHASE 7: CHURN RATE BY SIGNUP COHORT

# Keep customers with a known churn classification
classified_customers = customer_rfm[
    customer_rfm["churn_status"].notna()
].copy()

# Extract the year each customer signed up
classified_customers["signup_cohort"] = (
    classified_customers["signup_date"].dt.year
)

# Label missing signup years as Unknown
classified_customers["signup_cohort"] = (
    classified_customers["signup_cohort"]
    .astype("Int64")
    .astype("string")
    .fillna("Unknown")
)

# Group customers by signup cohort
churn_by_cohort = (
    classified_customers
    .groupby("signup_cohort")
    .agg(
        total_customers=("customer_id", "count"),
        churned_customers=("churn_status", "sum")
    )
    .reset_index()
)

# Calculate churn rate as a percentage
churn_by_cohort["churn_rate_pct"] = (
    churn_by_cohort["churned_customers"]
    / churn_by_cohort["total_customers"]
    * 100
).round(2)

# Display the results
print("Churn rate by signup cohort:")
display(churn_by_cohort)
```

```python
# PHASE 7: INSPECT SUPPORT TICKET COLUMNS

print("Support ticket columns:")
print(support_tickets.columns.tolist())

print("\nFirst five support ticket records:")
display(support_tickets.head())
```

```python
# PHASE 7: SUMMARISE CUSTOMER SUPPORT TICKETS

# Count the number of support tickets for each customer
ticket_count = (
    support_tickets.groupby("customer_id")
    .size()
    .reset_index(name="ticket_count")
)

# Calculate each customer's average satisfaction score
ticket_satisfaction = (
    support_tickets.groupby("customer_id")["satisfaction_score"]
    .mean()
    .reset_index(name="avg_satisfaction_score")
)

# Calculate each customer's average ticket resolution time
ticket_resolution = (
    support_tickets.groupby("customer_id")["resolution_time_hours"]
    .mean()
    .reset_index(name="avg_resolution_time_hours")
)

# Combine the three summaries
ticket_summary = (
    ticket_count
    .merge(ticket_satisfaction, on="customer_id", how="left")
    .merge(ticket_resolution, on="customer_id", how="left")
)

# Display the first five records
display(ticket_summary.head())

# Check the number of customers represented
print("Customers with support tickets:", ticket_summary["customer_id"].nunique())
print("Total support tickets:", support_tickets.shape[0])
```

Customer Support Summary

The analysis identified 5,200 support tickets associated with 3,490 distinct customers. For each customer, I calculated the number of support tickets, average satisfaction score, and average resolution time in hours.

These measures will be combined with the customer churn dataset to investigate whether support activity, satisfaction, and resolution time are associated with churn. The analysis will help identify potential opportunities to improve customer retention.

```python
# PHASE 7: MERGE CUSTOMER SUPPORT METRICS WITH CHURN DATA

# Combine the support summary with the customer churn dataset
customer_support = customer_rfm.merge(
    ticket_summary,
    on="customer_id",
    how="left"
)

# Customers without support tickets are assigned a ticket count of zero
customer_support["ticket_count"] = (
    customer_support["ticket_count"].fillna(0)
)

# Check that all customer records have been preserved
print("Customer records:", len(customer_support))
print("Customers with support tickets:",
      (customer_support["ticket_count"] > 0).sum())
print("Customers without support tickets:",
      (customer_support["ticket_count"] == 0).sum())

# Display a sample of the merged dataset
display(
    customer_support[
        ["customer_id", "churn_status", "ticket_count",
         "avg_satisfaction_score", "avg_resolution_time_hours"]
    ].head()
)
```

Merging Customer Support Metrics with Churn Data

I merged the customer-level support-ticket summary with the customer churn dataset using customer_id as the common key. A left join was used to preserve all 6,002 customer records.

The results showed that 3,490 customers had recorded support tickets, while 2,512 customers had no recorded support tickets. For customers without tickets, the ticket count was set to zero, while satisfaction scores and resolution times were left as missing because no support experience was recorded.

This combined dataset will be used to investigate the relationship between customer support experiences and churn status. Customers without a valid churn classification will be excluded from the comparison to ensure that the analysis uses only classifiable customers.

```python
# PHASE 7: COMPARE SUPPORT EXPERIENCES BY CHURN STATUS
# Exclude customers whose churn status is unknown
classified_support = customer_support[
    customer_support["churn_status"].notna()
].copy()
# Compare support metrics for churned and non-churned customers
support_comparison = (
    classified_support.groupby("churn_status")
    .agg(
        customer_count=("customer_id", "count"),
        avg_ticket_count=("ticket_count", "mean"),
        avg_satisfaction_score=("avg_satisfaction_score", "mean"),
        avg_resolution_time_hours=("avg_resolution_time_hours", "mean")
    )
    .reset_index()
)
# Replace numeric churn labels with readable descriptions
support_comparison["churn_status"] = (
    support_comparison["churn_status"]
    .map({0: "Not Churned", 1: "Churned"})
)
# Round the results for easier interpretation
support_comparison = support_comparison.round(2)
display(support_comparison)
```

Support Experiences and Customer Churn

I compared average support-ticket frequency, customer satisfaction, and ticket resolution time between churned and non-churned customers.

The analysis showed that churned customers averaged 0.88 support tickets, compared with 0.85 for non-churned customers. Their average satisfaction score was slightly lower at 3.86 compared with 3.91, while their average resolution time was higher at 18.66 hours compared with 17.37 hours.

These differences suggest possible associations between support experiences and customer churn. However, the differences are relatively small and do not establish causation. Further analysis is required to determine whether these measures provide meaningful insight into customer retention.

```python
# PHASE 7: ANALYSE SUPPORT TICKET ISSUE CATEGORIES

# Count tickets by issue category
issue_analysis = (
    support_tickets.groupby("issue_category")
    .agg(
        ticket_count=("ticket_id", "count"),
        avg_satisfaction=("satisfaction_score", "mean"),
        avg_resolution_hours=("resolution_time_hours", "mean")
    )
    .reset_index()
)

# Sort by the number of tickets
issue_analysis = issue_analysis.sort_values(
    by="ticket_count",
    ascending=False
)

# Round average values
issue_analysis[["avg_satisfaction", "avg_resolution_hours"]] = (
    issue_analysis[["avg_satisfaction", "avg_resolution_hours"]].round(2)
)

display(issue_analysis)
```

I analysed 5,200 support tickets across eight issue categories, comparing ticket frequency, average customer satisfaction, and average resolution time.

Product questions were the most frequent category, with 715 tickets, while wrong items shipped had the fewest, with 601 tickets. Billing issues had the longest average resolution time at 19.69 hours and the lowest average satisfaction score at 3.84. Product questions recorded the highest average satisfaction score of 3.94.

These findings suggest that billing support may be an area for operational improvement. Meridian could investigate billing-related processes and customer feedback to identify ways to improve resolution speed and satisfaction. Further analysis is needed to establish whether specific support issues are associated with customer churn.

```python
# PHASE 7: CALCULATE CHURN RATE BY SUPPORT ISSUE CATEGORY

# Attach churn status to support tickets
tickets_with_churn = support_tickets.merge(
    customer_support[["customer_id", "churn_status"]],
    on="customer_id",
    how="left"
)

# Keep customers with a known churn status
tickets_with_churn = tickets_with_churn[
    tickets_with_churn["churn_status"].notna()
].copy()

# Keep one record per customer per issue category
customer_issue = tickets_with_churn[
    ["customer_id", "issue_category", "churn_status"]
].drop_duplicates()

# Calculate churn by issue category
churn_by_issue = (
    customer_issue.groupby("issue_category")
    .agg(
        customers_with_issue=("customer_id", "nunique"),
        churned_customers=("churn_status", "sum")
    )
    .reset_index()
)

# Calculate the percentage of customers with each issue who churned
churn_by_issue["churn_rate_pct"] = (
    churn_by_issue["churned_customers"]
    / churn_by_issue["customers_with_issue"] * 100
).round(2)

# Display highest churn rates first
churn_by_issue = churn_by_issue.sort_values(
    "churn_rate_pct", ascending=False
)

display(churn_by_issue)
```

Churn Rate by Support Issue Category

I examined customer churn rates across eight support-ticket categories. Product Question had the highest observed churn rate at 66.19%, followed by Wrong Item Shipped at 65.98%. Delivery Delay had the lowest observed rate at 61.45%.

All eight categories recorded churn rates above 61%, indicating that churn affects customers across different support-issue categories. Meridian should investigate product information and fulfilment processes while continuing to improve customer retention more broadly.

These results show associations rather than causation. Customers may appear in multiple issue categories, and the analysis does not establish whether a support issue occurred before churn.
